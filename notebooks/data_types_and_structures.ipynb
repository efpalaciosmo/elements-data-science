```python
import polars as pl
```

**series**

```python
s = pl.Series("ints", [1, 2, 3, 4, 5])
print(s)
```

```python
s1 = pl.Series("ints", [1, 2, 3, 4, 5])
s2 = pl.Series("uints", [1, 2, 3, 4, 5], dtype=pl.UInt64)
print(s1.dtype, s2.dtype)
```

**Dataframes**

```python
from datetime import date

df = pl.DataFrame(
    {
        "name": ["Alice Archer", "Ben Brown", "Chloe Cooper", "Daniel Donovan"],
        "birthdate": [
            date(1997, 1, 10),
            date(1985, 2, 15),
            date(1983, 3, 22),
            date(1981, 4, 30),
        ],
        "weight": [57.9, 72.5, 53.6, 83.1],  # (kg)
        "height": [1.56, 1.77, 1.65, 1.75],  # (m)
    }
)

print(df)
```

`Glimpse`

Shows the values of the first few rows of a dataframe, each line of the output corresponds to a single column

```python
print(df.glimpse(return_type="string"))
```

`sample`

Its used to retrieve an arbitrary number of randomply selected rows from the dataframe.

```python
pl.set_random_seed(42)

print(df.sample(2))
```

`describe`
Summary statistics for all columns of the datafram

```python
print(df.describe())
```

`Schema`

Is a mapping of column or series name to the data types of those same columns or series

```python
print(df.schema)
```

We can specify an explicit schema by using a dictionary to map column names to data types.

```python
df = pl.DataFrame(
    {
        "name": ["Alice", "Ben", "Chloe", "Daniel"],
        "age": [27, 39, 41, 43],
    },
    schema={"name": str, "age": pl.UInt8}
)

print(df)
```
